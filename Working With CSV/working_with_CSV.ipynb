**14. Converters**

```python
# some time we are apply some feature and transformation at columns, even befour uploded the data set.
pd.read_csv("IPL Matches 2008-2020.csv")
# when we need the RCB at a place of "Royal Chalangers Bangalrow"
```

```python
# Then i creat a function 
def rename(name):
    if name == "Royal Challengers Bangalore":
        return "RCB"
    if name == "Kolkata Knight Riders":
        return "KKR"
    if name == "Chennai Super Kings":
        return "CSK"
    if name == "Rajasthan Royals":
        return "RR"
    if name == "Mumbai Indians":
        return "MI"
    else:
        return name
```

```python
rename( "Royal Challengers Bangalore")
# Then i can apply this transformation at IPL Matches 2008-2020.csv data set
```

```python
# pd.read_csv("IPL Matches 2008-2020.csv",converters={"team1":rename})
pd.read_csv('IPL Matches 2008-2020.csv',converters={'team1':rename,'team2':rename,'toss_winner':rename,'winner':rename})

```
