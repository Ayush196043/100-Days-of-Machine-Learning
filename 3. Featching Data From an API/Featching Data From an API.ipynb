```python
import time
import pandas as pd
import requests

# 1. Session use karenge taaki baar-baar naya connection na banana pade
session = requests.Session()

# 2. Saare pages ka data pehle list mein store karenge
dfs = []

api_key = "a8f59dd953b013260df12fd1b0aff412"
base_url = "https://api.themoviedb.org/3/movie/top_rated"

print("Data fetch hona shuru ho raha hai...")

for i in range(1, 551):
    # Har loop mein page number 'i' change hoga (1, 2, 3... 550)
    params = {"language": "en-US", "page": i, "api_key": "a8f59dd953b013260df12fd1b0aff412"}

    try:
        response = session.get(base_url, params=params)

        if response.status_code == 200:
            data = response.json().get("results", [])
            if not data:
                print(f"Page {i} par koi data nahi mila. Loop rok rahe hain.")
                break

            # Sirf kaam ke columns filter karein
            temp_df = pd.DataFrame(data)[
                [
                    "id",
                    "title",
                    "overview",
                    "popularity",
                    "release_date",
                    "vote_average",
                    "vote_count",
                ]
            ]
            dfs.append(temp_df)
        else:
            print(f"Page {i} par error aaya: Status Code {response.status_code}")

        # 3. SABSE ZAROORI: Har request ke baad 0.2 seconds ka gap (break) dein
        # Taaki server ko gussa na aaye aur connection reset na ho.
        time.sleep(0.2)

    except requests.exceptions.RequestException as e:
        print(f"Page {i} par network drop hua. Error: {e}")
        print("5 seconds ruk kar dobara try kar rahe hain...")
        time.sleep(5)
        continue

# 4. Aakhiri mein saari choti tables ko ek sath jod (concat) dein
if dfs:
    df = pd.concat(dfs, ignore_index=True)
    print(f"Mubarak ho! Total {len(df)} movies ka data mil gaya.")
else:
    df = pd.DataFrame()
    print("Koi data fetch nahi ho paya.")
```

```python
df.to_csv('movies_dataset.csv', index=False)
```
